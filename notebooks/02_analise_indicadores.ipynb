```python
perfil = df_analise['fe_enquadramento'].value_counts()

print("Perfil de Enquadramento:")
print(perfil)
print("\nTotal:", perfil.sum())
```

```python
percentuais = (perfil / perfil.sum()) * 100

print("Percentuais de Enquadramento:")
print(percentuais.round(2))
```

```python
tabela_perfil = pd.DataFrame({
    'Quantidade': perfil,
    'Percentual (%)': percentuais.round(2)
})

tabela_perfil
```

```python
tabela_perfil.index.name = 'Enquadramento'
tabela_perfil = tabela_perfil.reset_index()

tabela_perfil
```

```python
import matplotlib.pyplot as plt
grafico_perfil = tabela_perfil.copy()

grafico_perfil['Perfil'] = [
    'Docentes',
    'Técnico-administrativos',
    'Discentes de pós-graduação'
]

fig, ax = plt.subplots(figsize=(9, 3.5))

barras = ax.barh(
    grafico_perfil['Perfil'],
    grafico_perfil['Quantidade'],
    height=0.5,
    color='#2563EB'
)

for i, (_, linha) in enumerate(grafico_perfil.iterrows()):
    quantidade = linha['Quantidade']
    percentual = linha['Percentual (%)']

    ax.text(
        quantidade + 1,
        i,
        f'{quantidade} ({percentual:.1f}%)',
        va='center',
        fontsize=10
    )

ax.invert_yaxis()

ax.set_xlim(0, 82)
ax.set_xticks([])
ax.tick_params(axis='y', length=0)

for borda in ax.spines.values():
    borda.set_visible(False)

ax.set_title(
    'Perfil dos participantes — ICH/UFJF',
    loc='left',
    fontsize=14,
    fontweight='bold',
    pad=18
)

plt.tight_layout()
plt.show()
```


### Análise do Perfil dos Participantes

Dos 71 participantes da pesquisa, o grupo majoritário foi o dos docentes, com 67 respostas (94,4%). Em seguida, aparecem os técnicos-administrativos, com 3 respostas (4,2%), e os discentes de pós-graduação, com apenas 1 resposta (1,4%).

Essa distribuição deve ser considerada na interpretação dos resultados, pois a predominância dos docentes faz com que a pesquisa reflita principalmente a perspectiva desse grupo. A baixa participação dos técnicos-administrativos e discentes limita a possibilidade de generalizar os resultados para esses segmentos.


## Etapa 3 — Familiaridade com Inteligência Artificial

```python
df_analise['fe_familiaridade_ia'].unique()
```

```python
familiaridade = df_analise['fe_familiaridade_ia'].value_counts()

print("Familiaridade com IA:")
print(familiaridade)

print("\nTotal:", familiaridade.sum())
```

```python
percentuais_familiaridade = ((familiaridade / familiaridade.sum()) * 100).round(2)

print("Percentuais de Familiaridade com IA:")
print(percentuais_familiaridade)

print(f"Soma: {percentuais_familiaridade.sum():.2f}%")
```

```python
tabela_familiaridade = pd.DataFrame({
    'Quantidade': familiaridade,
    'Percentual (%)': percentuais_familiaridade
})

tabela_familiaridade.index.name = 'Familiaridade com IA'

tabela_familiaridade = tabela_familiaridade.reset_index()

tabela_familiaridade
```

```python
grafico_familiaridade = tabela_familiaridade.copy()
nomes = {
    'Utilizo com frequência para tarefas específicas': 'Uso frequente',
    'Já utilizei, mas muito pouco': 'Pouco uso',
    'Nunca utilizei': 'Nunca utilizei',
    'Utilizo diariamente e integro em meu fluxo de trabalho': 'Uso diário'
}

grafico_familiaridade['Categoria'] = (
    grafico_familiaridade['Familiaridade com IA'].replace(nomes)
)


fig, ax = plt.subplots(figsize=(9, 4))

barras = ax.barh(
    grafico_familiaridade['Categoria'],
    grafico_familiaridade['Quantidade'],
    color='#2563EB',
    height=0.55
)

rotulos = [
    f'{q} ({p:.1f}%)'
    for q, p in zip(
        grafico_familiaridade['Quantidade'],
        grafico_familiaridade['Percentual (%)']
    )
]

ax.bar_label(barras, labels=rotulos, padding=6)

# Ajustes visuais
ax.invert_yaxis()
ax.set_xlim(0, 45)
ax.set_xticks([])
ax.tick_params(axis='y', length=0)

for borda in ax.spines.values():
    borda.set_visible(False)

ax.set_title(
    'Familiaridade com IA — ICH/UFJF',
    loc='left',
    fontsize=14,
    fontweight='bold',
    pad=18
)

plt.tight_layout()
plt.show()
```

### Análise da Familiaridade com Inteligência Artificial

Dos 71 participantes da pesquisa, a resposta predominante foi "Utilizo com frequência para tarefas específicas", com 36 respostas (50,7%). Em seguida, aparece "Já utilizei, mas muito pouco", com 25 respostas (35,2%). As categorias "Nunca utilizei" e "Utilizo diariamente e integro em meu fluxo de trabalho" apresentaram 5 respostas cada (7,0%).

Observa-se que 41 participantes (57,7%) declararam utilizar IA frequentemente ou diariamente, indicando que essas ferramentas já fazem parte das atividades de uma parcela expressiva dos respondentes. Entretanto, 30 participantes (42,3%) relataram pouca ou nenhuma utilização, mostrando que ainda existem diferenças significativas na experiência com IA entre os participantes da pesquisa.

## Etapa 4 — Usos de Inteligência Artificial na Pesquisa

```python
colunas_uso = [
    coluna
    for coluna in df_analise.columns
    if coluna.startswith('uso_')
]

for coluna in colunas_uso:
    print(coluna)

print('\nTotal de categorias:', len(colunas_uso))
```

```python
contagem_uso = df_analise[colunas_uso].sum().sort_values(ascending=False)
print("Contagem de uso das categorias de IA:")
print(contagem_uso)
```

```python
percentuais_uso = (
    contagem_uso / len(df_analise) * 100
).round(2)

print("Percentuais dos usos de IA:")
print(percentuais_uso)
```

```python
tabela_usos = pd.DataFrame({
    'Quantidade': contagem_uso,
    'Percentual (%)': percentuais_uso
})

tabela_usos.index.name = 'Uso de IA'

tabela_usos = tabela_usos.reset_index()

tabela_usos
```

```python
from matplotlib.patches import Patch

grafico_usos = tabela_usos.copy()
nomes_usos = {
    'uso_traducao_textos': 'Tradução de textos',
    'uso_levantamento_bibliografico': 'Levantamento bibliográfico',
    'uso_revisao_textos': 'Revisão de textos',
    'uso_sintese_textos': 'Síntese de textos',
    'uso_transcricao': 'Transcrição',
    'uso_revisao_bibliografica': 'Revisão bibliográfica',
    'uso_geracao_ideias': 'Geração de ideias',
    'uso_analise_dados': 'Análise de dados',
    'uso_nunca_utilizei': 'Nunca utilizei',
    'uso_redacao_parcial': 'Redação parcial',
    'uso_redacao_total': 'Redação total',
    'uso_nao_se_aplica': 'Não se aplica',
    'uso_outros': 'Outros — respostas abertas',
}
grafico_usos['Categoria'] = grafico_usos['Uso de IA'].replace(nomes_usos)
assert len(grafico_usos) == 13
assert grafico_usos['Categoria'].str.startswith('uso_').sum() == 0

tabela_outros_usos = pd.read_csv(raiz / 'data' / 'resumo_outros_usos.csv')
assert tabela_outros_usos['Menções'].sum() == 12
assert len(tabela_outros_usos) == 7

principal = grafico_usos[['Categoria', 'Quantidade', 'Percentual (%)']].copy()
principal['Origem'] = 'Alternativas do formulário'

complemento = tabela_outros_usos.rename(
    columns={'Tema': 'Categoria', 'Menções': 'Quantidade'}
).copy()
complemento['Categoria'] = '    ↳ ' + complemento['Categoria']
complemento['Percentual (%)'] = float('nan')
complemento['Origem'] = 'Temas de Outros'

dados_grafico = pd.concat([principal, complemento], ignore_index=True)
cores = [
    '#F59E0B' if categoria == 'Outros — respostas abertas'
    else '#2563EB' if origem == 'Alternativas do formulário'
    else '#64748B'
    for categoria, origem in zip(dados_grafico['Categoria'], dados_grafico['Origem'])
]

fig, ax = plt.subplots(figsize=(13, 10))
barras = ax.barh(dados_grafico['Categoria'], dados_grafico['Quantidade'],
                 color=cores, height=0.64)
rotulos = [
    f"{int(linha['Quantidade'])} ({linha['Percentual (%)']:.1f}%)"
    if linha['Origem'] == 'Alternativas do formulário'
    else f"{int(linha['Quantidade'])} {'menção' if int(linha['Quantidade']) == 1 else 'menções'}"
    for _, linha in dados_grafico.iterrows()
]
ax.bar_label(barras, labels=rotulos, padding=6, fontsize=9)
ax.axhline(len(principal) - 0.5, color='#CBD5E1', linestyle='--', linewidth=1)
ax.invert_yaxis()
ax.set_xlim(0, max(1, dados_grafico['Quantidade'].max()) * 1.25 + 2)
ax.set_xticks([])
ax.tick_params(axis='y', length=0, labelsize=9)
for borda in ax.spines.values():
    borda.set_visible(False)

ax.legend(handles=[
    Patch(color='#2563EB', label='Alternativas originais'),
    Patch(color='#F59E0B', label='Outros — respondentes'),
    Patch(color='#64748B', label='Temas de Outros — menções'),
], loc='lower right', frameon=False, fontsize=9)
ax.set_title('Usos de Inteligência Artificial na pesquisa — ICH/UFJF',
             loc='left', fontsize=14, fontweight='bold', pad=18)
fig.text(0.02, 0.01,
         'Base: 71 respondentes. Alternativas: pessoas (%). Temas de Outros: menções; não somar as barras.',
         fontsize=9, color='#475569')
plt.tight_layout(rect=[0, 0.035, 1, 1])
plt.show()
```

### Análise dos Usos de Inteligência Artificial na Pesquisa

Dos 71 participantes da pesquisa, a atividade mais assinalada foi a tradução de textos, com 41 marcações (57,8%). Em seguida, aparecem o levantamento bibliográfico, com 36 marcações (50,7%), e a revisão de textos, com 29 marcações (40,9%). As categorias "Redação total de seções" e "Não se aplica" não receberam nenhuma marcação.

Observa-se que a utilização de IA para a redação de textos foi pouco frequente entre os participantes, com apenas 2 marcações (2,8%) para redação parcial e nenhuma para redação total.

Esses resultados sugerem que o uso declarado de IA na pesquisa está mais concentrado em atividades de apoio, como tradução, levantamento bibliográfico e revisão de textos, do que na produção direta de conteúdo acadêmico.

**Complementação — respostas de "Outros":** oito participantes (11,3%) tiveram respostas abertas classificadas nessa opção. Essas respostas somaram 12 menções temáticas, agrupadas em sete temas. Os mais frequentes foram formatação e produção visual/audiovisual (3 menções cada), seguidos de normalização bibliográfica (2 menções). As menções são um detalhamento de "Outros", não novos participantes, e podem coexistir com alternativas fechadas.


## Etapa 5 — Riscos do uso de IA na Gestão e Administração

```python
colunas_riscos = [
    coluna
    for coluna in df_analise.columns
    if coluna.startswith('riscos_')
]

for coluna in colunas_riscos:
    print(coluna)

print('\nTotal de categorias:', len(colunas_riscos))
```

```python
contagem_riscos = (
    df_analise[colunas_riscos]
    .sum()
    .sort_values(ascending=False)
)

print("Contagem dos riscos relacionados à IA:")
print(contagem_riscos)
```

```python
percentuais_riscos = (
    contagem_riscos / len(df_analise) * 100
).round(2)

print("Percentuais dos riscos relacionados à IA:")
print(percentuais_riscos)
```

```python
tabela_riscos = pd.DataFrame({
    'Quantidade': contagem_riscos,
    'Percentual (%)': percentuais_riscos
})

tabela_riscos.index.name = 'Riscos relacionados à IA'

tabela_riscos = tabela_riscos.reset_index()

tabela_riscos
```

```python
from matplotlib.patches import Patch

grafico_riscos = tabela_riscos.copy()
nomes_riscos = {
    'riscos_informacoes_enviesadas': 'Informações enviesadas',
    'riscos_violacoes_sigilo': 'Violações de sigilo',
    'riscos_precarizacao_trabalho': 'Precarização do trabalho',
    'riscos_sem_riscos': 'Sem riscos percebidos',
    'riscos_outros': 'Outros — respostas abertas',
}
grafico_riscos['Categoria'] = grafico_riscos['Riscos relacionados à IA'].replace(nomes_riscos)
assert len(grafico_riscos) == 5
assert grafico_riscos['Categoria'].str.startswith('riscos_').sum() == 0

tabela_outros_riscos = pd.read_csv(raiz / 'data' / 'resumo_outros_riscos.csv')
assert len(tabela_outros_riscos) == 5
assert tabela_outros_riscos['Menções'].sum() == 5

principal = grafico_riscos[['Categoria', 'Quantidade', 'Percentual (%)']].copy()
principal['Origem'] = 'Alternativas do formulário'
complemento = tabela_outros_riscos.rename(
    columns={'Tema': 'Categoria', 'Menções': 'Quantidade'}
).copy()
complemento['Categoria'] = '    ↳ ' + complemento['Categoria']
complemento['Percentual (%)'] = float('nan')
complemento['Origem'] = 'Temas de Outros'

dados_riscos = pd.concat([principal, complemento], ignore_index=True)
cores = [
    '#F59E0B' if categoria == 'Outros — respostas abertas'
    else '#2563EB' if origem == 'Alternativas do formulário'
    else '#64748B'
    for categoria, origem in zip(dados_riscos['Categoria'], dados_riscos['Origem'])
]

fig, ax = plt.subplots(figsize=(13, 6.5))
barras = ax.barh(dados_riscos['Categoria'], dados_riscos['Quantidade'],
                 color=cores, height=0.62)
rotulos = [
    f"{int(linha['Quantidade'])} ({linha['Percentual (%)']:.1f}%)"
    if linha['Origem'] == 'Alternativas do formulário'
    else f"{int(linha['Quantidade'])} {'menção' if int(linha['Quantidade']) == 1 else 'menções'}"
    for _, linha in dados_riscos.iterrows()
]
ax.bar_label(barras, labels=rotulos, padding=6, fontsize=9)
ax.axhline(len(principal) - 0.5, color='#CBD5E1', linestyle='--', linewidth=1)
ax.invert_yaxis()
ax.set_xlim(0, max(1, dados_riscos['Quantidade'].max()) * 1.25 + 2)
ax.set_xticks([])
ax.tick_params(axis='y', length=0, labelsize=9)
for borda in ax.spines.values():
    borda.set_visible(False)

ax.legend(handles=[
    Patch(color='#2563EB', label='Alternativas originais'),
    Patch(color='#F59E0B', label='Outros — respondentes'),
    Patch(color='#64748B', label='Temas de Outros — menções'),
], loc='lower right', frameon=False, fontsize=9)
ax.set_title('Riscos do uso de IA na gestão e administração — ICH/UFJF',
             loc='left', fontsize=14, fontweight='bold', pad=18)
fig.text(0.02, 0.01,
         'Base: 71 respondentes. Alternativas: pessoas (%). Temas de Outros: menções; não somar as barras.',
         fontsize=9, color='#475569')
plt.tight_layout(rect=[0, 0.035, 1, 1])
plt.show()
```

### Análise dos Riscos do Uso de IA na Gestão e Administração

Dos 71 participantes da pesquisa, o risco mais assinalado foi o de informações enviesadas, com 54 marcações (76,1%). Em seguida, aparecem as violações de sigilo, com 36 marcações (50,7%), e a precarização do trabalho, com 32 marcações (45,1%). Apenas 5 participantes (7,0%) assinalaram a alternativa "Sem riscos percebidos".

Além das alternativas originais, 5 participantes (7,0%) apresentaram respostas na opção "Outros". Essas respostas foram categorizadas em cinco temas: segurança, governança e conformidade; crítica genérica à IA; superficialidade das informações; necessidade de revisão humana; e monitoramento contínuo de riscos. Cada tema apresentou uma menção.

Os resultados indicam uma preocupação expressiva dos participantes com os possíveis impactos da IA na gestão e administração, principalmente em relação à confiabilidade das informações, à proteção de dados sigilosos e às condições de trabalho.

As respostas abertas complementam essas preocupações, destacando também a importância da supervisão humana e da adoção de mecanismos de controle e monitoramento.

Por fim, é importante considerar que os resultados representam percepções dos respondentes, e não ocorrências comprovadas desses riscos. Como a pergunta permitia múltiplas escolhas, os percentuais não precisam somar 100%.
