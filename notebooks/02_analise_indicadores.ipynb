## Etapa 6 — Opiniões sobre a Portaria do CNPq

```python
opinioes_portaria = df_analise['portaria_cnpq_categoria']

print("Opiniões sobre a Portaria do CNPq:")
print(opinioes_portaria.value_counts(dropna=False))

print("\nTotal de participantes:", len(df_analise))
print("Respostas válidas:", opinioes_portaria.notna().sum())
print("Respostas ausentes:", opinioes_portaria.isna().sum())
```

```python
categorias_originais = [
    'Muito adequada',
    'Adequada, mas com desafios de implementação',
    'Restritiva demais para a realidade das Humanidades',
    'Frágil, pois depende muito da honestidade do pesquisador',
    'Não tenho opinião formada',
    'Desconheço as diretrizes'
]

outros_portaria = opinioes_portaria[
    opinioes_portaria.notna()
    & ~opinioes_portaria.isin(categorias_originais)
]

print("Categorias classificadas como Outros:")
print(outros_portaria.value_counts())

print("\nTotal de respostas em Outros:", len(outros_portaria))
```

```python
situacao_resposta = (
    df_analise['portaria_cnpq_categoria']
    .isna()
    .map({
        True: 'Não respondeu',
        False: 'Respondeu'
    })
)

tabela_ausencias = pd.crosstab(
    df_analise['fe_enquadramento'],
    situacao_resposta
)

print("Respostas sobre a Portaria por perfil:")
display(tabela_ausencias)
```

```python
portaria_validas = opinioes_portaria.dropna()

contagem_portaria = (
    portaria_validas
    .value_counts()
)


percentuais_portaria = (
    contagem_portaria / len(portaria_validas) * 100
).round(2)

print("Percentuais das opiniões sobre a Portaria:")
print(percentuais_portaria)

print("\nTotal de respostas válidas:", len(portaria_validas))
```

```python
ordem_portaria = categorias_originais + [
    categoria
    for categoria in contagem_portaria.index
    if categoria not in categorias_originais
]

tabela_portaria = pd.DataFrame({
    'Quantidade': contagem_portaria.reindex(
        ordem_portaria,
        fill_value=0
    )
})

tabela_portaria['Percentual (%)'] = (
    tabela_portaria['Quantidade'] / len(portaria_validas) * 100
).round(2)

tabela_portaria.index.name = 'Opinião sobre a Portaria'
tabela_portaria = tabela_portaria.reset_index()

display(tabela_portaria)

print("Total de respostas:", tabela_portaria['Quantidade'].sum())
print("Total de categorias:", len(tabela_portaria))
```

```python

from matplotlib.patches import Patch

originais = tabela_portaria.iloc[:len(categorias_originais)].copy()
outros = tabela_portaria.iloc[len(categorias_originais):].copy()

originais = originais.sort_values(
    by='Quantidade',
    ascending=False
)

outros = outros.sort_values(
    by='Quantidade',
    ascending=False
)

originais['Tipo'] = 'Original'
outros['Tipo'] = 'Detalhamento'

total_outros = outros['Quantidade'].sum()

resumo_outros = pd.DataFrame({
    'Opinião sobre a Portaria': ['Outros — respostas abertas'],
    'Quantidade': [total_outros],
    'Percentual (%)': [
        round(total_outros / len(portaria_validas) * 100, 2)
    ],
    'Tipo': ['Outros']
})

grafico_portaria = pd.concat(
    [originais, resumo_outros, outros],
    ignore_index=True
)

nomes_portaria = {
    'Restritiva demais para a realidade das Humanidades':
        'Restritiva para as Humanidades',
    'Frágil, pois depende muito da honestidade do pesquisador':
        'Frágil: depende da honestidade do pesquisador'
}

grafico_portaria['Categoria'] = (
    grafico_portaria['Opinião sobre a Portaria']
    .replace(nomes_portaria)
)

cores = {
    'Original': '#2563EB',
    'Outros': '#F59E0B',
    'Detalhamento': '#64748B'
}

fig, ax = plt.subplots(figsize=(12, 6))

barras = ax.barh(
    grafico_portaria['Categoria'],
    grafico_portaria['Quantidade'],
    color=grafico_portaria['Tipo'].map(cores),
    height=0.6
)

rotulos = []

for _, linha in grafico_portaria.iterrows():
    quantidade = linha['Quantidade']

    if linha['Tipo'] == 'Detalhamento':
        rotulos.append(f'{quantidade} resposta')
    else:
        percentual = linha['Percentual (%)']
        rotulos.append(f'{quantidade} ({percentual:.1f}%)')

ax.bar_label(barras, labels=rotulos, padding=6)

ax.axhline(
    y=len(categorias_originais) + 0.5,
    color='#CBD5E1',
    linestyle='--'
)

ax.invert_yaxis()
ax.set_xlim(0, 40)
ax.set_xticks([])
ax.tick_params(axis='y', length=0)

for borda in ax.spines.values():
    borda.set_visible(False)

ax.legend(
    handles=[
        Patch(color='#2563EB', label='Alternativas originais'),
        Patch(color='#F59E0B', label='Outros — participantes'),
        Patch(color='#64748B', label='Categorias de Outros')
    ],
    loc='lower right',
    frameon=False
)

ax.set_title(
    'Opiniões sobre a Portaria do CNPq — ICH/UFJF',
    loc='left',
    fontsize=14,
    fontweight='bold',
    pad=18
)

plt.tight_layout()
plt.show()
```

### Análise das Opiniões sobre a Portaria do CNPq

Dos 68 participantes que responderam à pergunta sobre a Portaria do CNPq, a opinião predominante foi "Adequada, mas com desafios de implementação", com 30 respostas (44,1%). Em seguida, aparece "Muito adequada", com 19 respostas (27,9%). Juntas, essas duas categorias representam 49 participantes (72,1%), indicando uma avaliação predominantemente favorável à norma entre os respondentes.

Por outro lado, 9 participantes (13,2%) declararam não ter opinião formada, 5 (7,4%) consideraram a norma frágil por depender da honestidade do pesquisador e 3 (4,4%) afirmaram desconhecer as diretrizes. A alternativa "Restritiva demais para a realidade das Humanidades" não recebeu nenhuma marcação.

```python
raiz = Path.cwd()

if raiz.name.lower() == 'notebooks':
    raiz = raiz.parent

arquivo_etica = raiz / 'data' / 'categorias_etica.csv'

df_etica = pd.read_csv(arquivo_etica, encoding='utf-8-sig')

df_etica['Categorias'] = (
    df_etica['Categorias']
    .fillna('')
    .str.split('|')
    .apply(lambda categorias: [
        categoria.strip()
        for categoria in categorias
        if categoria.strip()
    ])
)

assert len(df_etica) == 71
assert df_etica['ID'].is_unique
assert df_etica['Categorias'].apply(len).min() > 0

print("Respostas importadas:", len(df_etica))
print(
    "Total de atribuições temáticas:",
    df_etica['Categorias'].apply(len).sum()
)

frequencias_etica = (
    df_etica['Categorias']
    .explode()
    .value_counts()
    .rename_axis('Categoria')
    .reset_index(name='Participantes')
)

frequencias_etica['Percentual'] = (
    frequencias_etica['Participantes']
    / len(df_etica) * 100
).round(1)

display(frequencias_etica)
```

Outro benefício recorrente foi a **síntese e revisão textual**, identificada em 18 respostas (26,5%), seguida pela **análise de dados**, mencionada por 14 participantes (20,6%). Essas categorias envolvem tarefas como elaboração de resumos, revisão de textos, processamento de informações e apoio à exploração de conjuntos de dados.

Também foram identificados benefícios relacionados ao **apoio operacional e administrativo**, mencionado por 11 participantes (16,2%), à utilização de **recursos visuais** e ao **apoio à formulação de ideias**, ambos presentes em 7 respostas (10,3%).

Entre os temas menos frequentes, apareceram a **formatação de trabalhos**, com 5 respostas (7,4%), o **foco na análise crítica** e a **ampliação metodológica e inovação**, com 4 respostas cada (5,9%), além da **verificação e qualidade das informações**, identificada em 3 respostas (4,4%).

#### Respostas sem benefício definido

Três participantes (4,4%) não identificaram benefícios específicos ou declararam não possuir opinião formada sobre o assunto.

Essas respostas foram classificadas separadamente, pois não representam benefícios concretos, mas manifestações de desconhecimento ou ausência de posicionamento definido.

#### Considerações gerais

Os resultados sugerem que os participantes percebem os principais benefícios da Inteligência Artificial nas **atividades de apoio e organização do trabalho acadêmico**, especialmente na economia de tempo, no levantamento bibliográfico, na organização de informações e no processamento de textos e dados.

Um aspecto relevante é que algumas respostas destacaram a possibilidade de utilizar o tempo economizado em tarefas operacionais para desenvolver atividades intelectuais mais complexas, como interpretação, reflexão crítica e elaboração de problemas de pesquisa.

Entretanto, os benefícios identificados representam as percepções dos respondentes e não comprovam, por si só, ganhos efetivos de produtividade ou qualidade científica.

### 7.9 — Diretrizes de Transparência e Controle na Gestão do ICH

```python
arquivo_transparencia = (
    raiz / 'data' / 'categorias_transparencia.csv'
)

df_transparencia = pd.read_csv(
    arquivo_transparencia,
    encoding='utf-8-sig'
)

df_transparencia['Categorias'] = (
    df_transparencia['Categorias']
    .fillna('')
    .str.split('|')
    .apply(lambda lista: [
        categoria.strip()
        for categoria in lista
        if categoria.strip()
    ])
)

assert len(df_transparencia) == 47
assert df_transparencia['ID'].is_unique
assert df_transparencia['Categorias'].apply(len).min() > 0

print("Respostas importadas:", len(df_transparencia))

print(
    "Total de atribuições temáticas:",
    df_transparencia['Categorias'].apply(len).sum()
)
```

```python
frequencias_transparencia = (
    df_transparencia['Categorias']
    .explode()
    .value_counts()
    .rename_axis('Categoria')
    .reset_index(name='Participantes')
)

frequencias_transparencia['Percentual'] = (
    frequencias_transparencia['Participantes']
    / len(df_transparencia) * 100
).round(1)

print("Total de categorias:", len(frequencias_transparencia))

print(
    "Total de atribuições temáticas:",
    frequencias_transparencia['Participantes'].sum()
)

display(frequencias_transparencia)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.patches import Patch

AZUL = '#2563EB'
CINZA = '#64748B'

categorias_posicionamento = [
    'Sem opinião ou sugestão',
    'Críticas e dificuldades de controle',
    'Uso operacional sem diretriz específica'
]

df_diretrizes = frequencias_transparencia[
    ~frequencias_transparencia['Categoria'].isin(
        categorias_posicionamento
    )
].copy()

df_posicionamentos = frequencias_transparencia[
    frequencias_transparencia['Categoria'].isin(
        categorias_posicionamento
    )
].copy()

df_diretrizes = df_diretrizes.sort_values(
    'Participantes', ascending=False
)

df_posicionamentos = df_posicionamentos.sort_values(
    'Participantes', ascending=False
)

df_grafico_transparencia = pd.concat(
    [df_diretrizes, df_posicionamentos],
    ignore_index=True
)

assert len(df_grafico_transparencia) == 14
assert df_grafico_transparencia['Participantes'].sum() == 81

cores = (
    [AZUL] * len(df_diretrizes)
    + [CINZA] * len(df_posicionamentos)
)

fig, ax = plt.subplots(figsize=(13, 9))

barras = ax.barh(
    df_grafico_transparencia['Categoria'],
    df_grafico_transparencia['Participantes'],
    color=cores,
    height=0.68
)

ax.invert_yaxis()

for barra, (_, linha) in zip(
    barras,
    df_grafico_transparencia.iterrows()
):
    quantidade = linha['Participantes']
    percentual = linha['Percentual']

    ax.text(
        quantidade + 0.3,
        barra.get_y() + barra.get_height() / 2,
        f'{quantidade} ({percentual:.1f}%)',
        va='center',
        fontsize=10
    )

ax.axhline(
    y=len(df_diretrizes) - 0.5,
    color='#94A3B8',
    linestyle='--',
    linewidth=1
)

ax.set_title(
    'Diretrizes de Transparência e Controle da IA na Gestão do ICH',
    fontsize=15,
    fontweight='bold',
    pad=20
)

ax.set_xlabel(
    'Número de participantes',
    fontsize=11
)

ax.set_xlim(
    0,
    df_grafico_transparencia['Participantes'].max() + 10
)

ax.grid(
    axis='x',
    linestyle='--',
    alpha=0.2
)

ax.set_axisbelow(True)

for lado in ['top', 'right', 'left']:
    ax.spines[lado].set_visible(False)

ax.tick_params(
    axis='y',
    length=0,
    labelsize=10
)

ax.legend(
    handles=[
        Patch(
            color=AZUL,
            label='Diretrizes e mecanismos sugeridos'
        ),
        Patch(
            color=CINZA,
            label='Outros posicionamentos'
        )
    ],
    loc='lower right',
    frameon=False,
    fontsize=9
)

fig.text(
    0.10,
    0.025,
    'Base: 47 respostas preenchidas; 24 ausentes entre 71 participantes. '
    'Categorias não exclusivas.',
    fontsize=9,
    color=CINZA
)

plt.tight_layout(rect=[0, 0.06, 1, 1])
plt.show()
```

### Análise das Diretrizes de Transparência e Controle da IA na Gestão do ICH

A análise das 47 respostas preenchidas identificou 14 categorias e 81 atribuições temáticas. A categoria mais frequente foi **"Sem opinião ou sugestão"**, com 16 participantes (34,0%). Além disso, 24 dos 71 participantes deixaram a pergunta sem resposta.

Entre as propostas apresentadas, destacaram-se a **declaração e transparência do uso de IA**, com 10 respostas (21,3%), a criação de **normas e políticas institucionais**, com 9 (19,1%), e a **supervisão e responsabilidade humana**, com 8 (17,0%). Também apareceram sugestões de **auditoria e fiscalização**, com 7 respostas (14,9%), e **proteção de dados e sigilo**, com 6 (12,8%).

Duas respostas (4,3%) foram classificadas como **"Críticas e dificuldades de controle"**, por abordarem limitações na fiscalização e na identificação do uso de IA. Outra resposta (2,1%) foi classificada como **"Uso operacional sem diretriz específica"**, pois mencionou a estruturação de dados sem propor mecanismos de transparência ou controle.

De modo geral, os resultados destacam a importância de estabelecer regras claras, garantir a transparência e preservar a supervisão humana na gestão do ICH.

**Nota metodológica:** Os percentuais consideram as 47 respostas preenchidas. As categorias não são exclusivas, e uma mesma resposta pode apresentar diferentes sugestões. As classificações em cinza representam manifestações que não constituem diretamente propostas de governança.
