```python
import pandas as pd
from pathlib import Path

raiz = Path.cwd()

if raiz.name.lower() == 'notebooks':
    raiz = raiz.parent

print("Pandas:", pd.__version__)
print("CSV encontrado:", (raiz / 'data' / 'ia.csv').exists())
print("JSON encontrado:", (raiz / 'data' / 'mapeamentos_categorias.json').exists())
```

```python
df_analise = pd.read_csv(raiz / 'data' / 'indicadores_ich.csv')

print("Dimensões do DataFrame:", df_analise.shape)
print("Colunas do DataFrame:", len(df_analise.columns))
print("Total de Registros:", len(df_analise))
```

```python
df_analise.info()
```

```python
perfil = df_analise['fe_enquadramento'].value_counts()

print("Perfil de Enquadramento:")
print(perfil)
print("\nTotal:", perfil.sum())
```

```python
percentuais = (perfil / perfil.sum()) * 100

print("Percentuais de Enquadramento:")
print(percentuais.round(2))
```

```python
tabela_perfil = pd.DataFrame({
    'Quantidade': perfil,
    'Percentual (%)': percentuais.round(2)
})

tabela_perfil
```

```python
tabela_perfil.index.name = 'Enquadramento'
tabela_perfil = tabela_perfil.reset_index()

tabela_perfil
```

```python
import matplotlib.pyplot as plt
grafico_perfil = tabela_perfil.copy()

grafico_perfil['Perfil'] = [
    'Docentes',
    'Técnico-administrativos',
    'Discentes de pós-graduação'
]

fig, ax = plt.subplots(figsize=(9, 3.5))

barras = ax.barh(
    grafico_perfil['Perfil'],
    grafico_perfil['Quantidade'],
    height=0.5,
    color='#2563EB'
)

for i, (_, linha) in enumerate(grafico_perfil.iterrows()):
    quantidade = linha['Quantidade']
    percentual = linha['Percentual (%)']

    ax.text(
        quantidade + 1,
        i,
        f'{quantidade} ({percentual:.1f}%)',
        va='center',
        fontsize=10
    )

ax.invert_yaxis()

ax.set_xlim(0, 82)
ax.set_xticks([])
ax.tick_params(axis='y', length=0)

for borda in ax.spines.values():
    borda.set_visible(False)

ax.set_title(
    'Perfil dos participantes — ICH/UFJF',
    loc='left',
    fontsize=14,
    fontweight='bold',
    pad=18
)

plt.tight_layout()
plt.show()
```


### Análise do Perfil dos Participantes

Dos 71 participantes da pesquisa, o grupo majoritário foi o dos docentes, com 67 respostas (94,4%). Em seguida, aparecem os técnicos-administrativos, com 3 respostas (4,2%), e os discentes de pós-graduação, com apenas 1 resposta (1,4%).

Essa distribuição deve ser considerada na interpretação dos resultados, pois a predominância dos docentes faz com que a pesquisa reflita principalmente a perspectiva desse grupo. A baixa participação dos técnicos-administrativos e discentes limita a possibilidade de generalizar os resultados para esses segmentos.
