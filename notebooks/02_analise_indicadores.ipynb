## Etapa 6 — Opiniões sobre a Portaria do CNPq

```python
opinioes_portaria = df_analise['portaria_cnpq_categoria']

print("Opiniões sobre a Portaria do CNPq:")
print(opinioes_portaria.value_counts(dropna=False))

print("\nTotal de participantes:", len(df_analise))
print("Respostas válidas:", opinioes_portaria.notna().sum())
print("Respostas ausentes:", opinioes_portaria.isna().sum())
```

```python
categorias_originais = [
    'Muito adequada',
    'Adequada, mas com desafios de implementação',
    'Restritiva demais para a realidade das Humanidades',
    'Frágil, pois depende muito da honestidade do pesquisador',
    'Não tenho opinião formada',
    'Desconheço as diretrizes'
]

outros_portaria = opinioes_portaria[
    opinioes_portaria.notna()
    & ~opinioes_portaria.isin(categorias_originais)
]

print("Categorias classificadas como Outros:")
print(outros_portaria.value_counts())

print("\nTotal de respostas em Outros:", len(outros_portaria))
```

```python
situacao_resposta = (
    df_analise['portaria_cnpq_categoria']
    .isna()
    .map({
        True: 'Não respondeu',
        False: 'Respondeu'
    })
)

tabela_ausencias = pd.crosstab(
    df_analise['fe_enquadramento'],
    situacao_resposta
)

print("Respostas sobre a Portaria por perfil:")
display(tabela_ausencias)
```

```python
portaria_validas = opinioes_portaria.dropna()

contagem_portaria = (
    portaria_validas
    .value_counts()
)


percentuais_portaria = (
    contagem_portaria / len(portaria_validas) * 100
).round(2)

print("Percentuais das opiniões sobre a Portaria:")
print(percentuais_portaria)

print("\nTotal de respostas válidas:", len(portaria_validas))
```

```python
ordem_portaria = categorias_originais + [
    categoria
    for categoria in contagem_portaria.index
    if categoria not in categorias_originais
]

tabela_portaria = pd.DataFrame({
    'Quantidade': contagem_portaria.reindex(
        ordem_portaria,
        fill_value=0
    )
})

tabela_portaria['Percentual (%)'] = (
    tabela_portaria['Quantidade'] / len(portaria_validas) * 100
).round(2)

tabela_portaria.index.name = 'Opinião sobre a Portaria'
tabela_portaria = tabela_portaria.reset_index()

display(tabela_portaria)

print("Total de respostas:", tabela_portaria['Quantidade'].sum())
print("Total de categorias:", len(tabela_portaria))
```

```python

from matplotlib.patches import Patch

originais = tabela_portaria.iloc[:len(categorias_originais)].copy()
outros = tabela_portaria.iloc[len(categorias_originais):].copy()

originais = originais.sort_values(
    by='Quantidade',
    ascending=False
)

outros = outros.sort_values(
    by='Quantidade',
    ascending=False
)

originais['Tipo'] = 'Original'
outros['Tipo'] = 'Detalhamento'

total_outros = outros['Quantidade'].sum()

resumo_outros = pd.DataFrame({
    'Opinião sobre a Portaria': ['Outros — respostas abertas'],
    'Quantidade': [total_outros],
    'Percentual (%)': [
        round(total_outros / len(portaria_validas) * 100, 2)
    ],
    'Tipo': ['Outros']
})

grafico_portaria = pd.concat(
    [originais, resumo_outros, outros],
    ignore_index=True
)

nomes_portaria = {
    'Restritiva demais para a realidade das Humanidades':
        'Restritiva para as Humanidades',
    'Frágil, pois depende muito da honestidade do pesquisador':
        'Frágil: depende da honestidade do pesquisador'
}

grafico_portaria['Categoria'] = (
    grafico_portaria['Opinião sobre a Portaria']
    .replace(nomes_portaria)
)

cores = {
    'Original': '#2563EB',
    'Outros': '#F59E0B',
    'Detalhamento': '#64748B'
}

fig, ax = plt.subplots(figsize=(12, 6))

barras = ax.barh(
    grafico_portaria['Categoria'],
    grafico_portaria['Quantidade'],
    color=grafico_portaria['Tipo'].map(cores),
    height=0.6
)

rotulos = []

for _, linha in grafico_portaria.iterrows():
    quantidade = linha['Quantidade']

    if linha['Tipo'] == 'Detalhamento':
        rotulos.append(f'{quantidade} resposta')
    else:
        percentual = linha['Percentual (%)']
        rotulos.append(f'{quantidade} ({percentual:.1f}%)')

ax.bar_label(barras, labels=rotulos, padding=6)

ax.axhline(
    y=len(categorias_originais) + 0.5,
    color='#CBD5E1',
    linestyle='--'
)

ax.invert_yaxis()
ax.set_xlim(0, 40)
ax.set_xticks([])
ax.tick_params(axis='y', length=0)

for borda in ax.spines.values():
    borda.set_visible(False)

ax.legend(
    handles=[
        Patch(color='#2563EB', label='Alternativas originais'),
        Patch(color='#F59E0B', label='Outros — participantes'),
        Patch(color='#64748B', label='Categorias de Outros')
    ],
    loc='lower right',
    frameon=False
)

ax.set_title(
    'Opiniões sobre a Portaria do CNPq — ICH/UFJF',
    loc='left',
    fontsize=14,
    fontweight='bold',
    pad=18
)

plt.tight_layout()
plt.show()
```

### Análise das Opiniões sobre a Portaria do CNPq

Dos 68 participantes que responderam à pergunta sobre a Portaria do CNPq, a opinião predominante foi "Adequada, mas com desafios de implementação", com 30 respostas (44,1%). Em seguida, aparece "Muito adequada", com 19 respostas (27,9%). Juntas, essas duas categorias representam 49 participantes (72,1%), indicando uma avaliação predominantemente favorável à norma entre os respondentes.

Por outro lado, 9 participantes (13,2%) declararam não ter opinião formada, 5 (7,4%) consideraram a norma frágil por depender da honestidade do pesquisador e 3 (4,4%) afirmaram desconhecer as diretrizes. A alternativa "Restritiva demais para a realidade das Humanidades" não recebeu nenhuma marcação.

```python
raiz = Path.cwd()

if raiz.name.lower() == 'notebooks':
    raiz = raiz.parent

arquivo_etica = raiz / 'data' / 'categorias_etica.csv'

df_etica = pd.read_csv(arquivo_etica, encoding='utf-8-sig')

df_etica['Categorias'] = (
    df_etica['Categorias']
    .fillna('')
    .str.split('|')
    .apply(lambda categorias: [
        categoria.strip()
        for categoria in categorias
        if categoria.strip()
    ])
)

assert len(df_etica) == 71
assert df_etica['ID'].is_unique
assert df_etica['Categorias'].apply(len).min() > 0

print("Respostas importadas:", len(df_etica))
print(
    "Total de atribuições temáticas:",
    df_etica['Categorias'].apply(len).sum()
)

frequencias_etica = (
    df_etica['Categorias']
    .explode()
    .value_counts()
    .rename_axis('Categoria')
    .reset_index(name='Participantes')
)

frequencias_etica['Percentual'] = (
    frequencias_etica['Participantes']
    / len(df_etica) * 100
).round(1)

display(frequencias_etica)
```

```python
atividades_uso = [
    coluna
    for coluna in colunas_uso
    if coluna not in [
        'uso_nunca_utilizei',
        'uso_nao_se_aplica'
    ]
]

df_cruzamentos['qtd_atividades_ia'] = (
    df_cruzamentos[atividades_uso]
    .fillna(False)
    .astype(bool)
    .sum(axis=1)
)

df_cruzamentos['utiliza_ia_pesquisa'] = (
    df_cruzamentos['qtd_atividades_ia'] > 0
)

ordem_familiaridade = [
    'Nunca utilizei',
    'Já utilizei, mas muito pouco',
    'Utilizo com frequência para tarefas específicas',
    'Utilizo diariamente e integro em meu fluxo de trabalho'
]

tabela_familiaridade_uso = pd.crosstab(
    df_cruzamentos['fe_familiaridade_ia'],
    df_cruzamentos['utiliza_ia_pesquisa']
).reindex(
    index=ordem_familiaridade,
    columns=[False, True],
    fill_value=0
)

tabela_familiaridade_uso.columns = [
    'Sem atividade informada',
    'Com atividade informada'
]

tabela_familiaridade_uso['Total'] = (
    tabela_familiaridade_uso.sum(axis=1)
)

tabela_familiaridade_uso['Percentual com atividade'] = (
    tabela_familiaridade_uso['Com atividade informada']
    / tabela_familiaridade_uso['Total'] * 100
).round(1)

display(tabela_familiaridade_uso)

print(
    "\nParticipantes com alguma atividade informada:",
    df_cruzamentos['utiliza_ia_pesquisa'].sum()
)

print(
    "Marcaram 'Nunca utilizei' nas atividades:",
    df_cruzamentos['uso_nunca_utilizei'].sum()
)

print(
    "Marcaram 'Nunca utilizei' e alguma atividade:",
    (
        df_cruzamentos['uso_nunca_utilizei']
        & df_cruzamentos['utiliza_ia_pesquisa']
    ).sum()
)
```

- **Orientações ou posicionamentos específicos:** incluem recomendações de uso ético e transparente, orientações baseadas em normas externas, políticas formalizadas e posicionamentos de proibição, restrição ou desestímulo ao uso.

De modo geral, os resultados reforçam a importância de estabelecer **diretrizes institucionais claras**, reduzindo a dependência de orientações informais e oferecendo maior segurança aos docentes e estudantes.

- **Sem riscos:** participantes que declararam não identificar riscos associados ao uso da tecnologia.

O risco mais mencionado nos dois grupos foi a geração de **informações enviesadas**, identificado por 73,3% dos participantes com baixo ou nenhum uso de IA e 78,0% daqueles com uso frequente ou diário. Essa diferença de 7 pontos percentuais indica uma preocupação ligeiramente maior com a proteção de informações entre os participantes que utilizam IA com maior frequência.

Em relação à **precarização do trabalho**, os resultados foram de 46,7% entre os participantes com baixo ou nenhum uso e 43,9% entre os usuários frequentes ou diários. Embora a diferença seja pequena, o resultado evidencia que a preocupação com os impactos da automação sobre o trabalho humano também está presente entre os dois grupos.

As categorias "Outros" e "Sem riscos" apresentaram frequências reduzidas, representando menos de 8% dos participantes em cada grupo.

De maneira geral, os resultados indicam que **a percepção dos riscos relacionados à IA não está restrita aos participantes com pouca familiaridade com a tecnologia**. ```python

```python
if 'Departamento_analise' in df_cruzamentos.columns:
    coluna_departamento = 'Departamento_analise'
elif 'fe_departamento' in df_cruzamentos.columns:
    coluna_departamento = 'fe_departamento'
else:
    raise ValueError(
        "Coluna de departamento não encontrada. "
        "Confira as colunas disponíveis."
    )

colunas_atividades = [
    'uso_levantamento_bibliografico',
    'uso_revisao_bibliografica',
    'uso_sintese_textos',
    'uso_analise_dados',
    'uso_traducao_textos',
    'uso_revisao_textos',
    'uso_geracao_ideias',
    'uso_redacao_parcial',
    'uso_redacao_total',
    'uso_transcricao',
    'uso_outros'
]

rotulos_atividades = {
    'uso_levantamento_bibliografico': 'Levantamento bibliográfico',
    'uso_revisao_bibliografica': 'Revisão bibliográfica',
    'uso_sintese_textos': 'Síntese de textos',
    'uso_analise_dados': 'Análise de dados',
    'uso_traducao_textos': 'Tradução de textos',
    'uso_revisao_textos': 'Revisão de textos',
    'uso_geracao_ideias': 'Geração de ideias',
    'uso_redacao_parcial': 'Redação parcial',
    'uso_redacao_total': 'Redação total',
    'uso_transcricao': 'Transcrição',
    'uso_outros': 'Outras atividades'
}

faltantes = [
    c for c in colunas_atividades
    if c not in df_cruzamentos.columns
]

if faltantes:
    raise ValueError(f"Colunas ausentes: {faltantes}")

df_depto_atividades = df_cruzamentos.copy()

def agrupar_departamento(valor):
    if pd.isna(valor):
        return np.nan

    nome = str(valor).strip()

    if nome.casefold() in ['filosofia', 'turismo']:
        return 'Filosofia e Turismo (agrupados)'

    return nome

df_depto_atividades['Departamento_analise'] = (
    df_depto_atividades[coluna_departamento]
    .apply(agrupar_departamento)
)

df_depto_atividades = df_depto_atividades.dropna(
    subset=['Departamento_analise']
).copy()

print("Participantes com departamento informado:",
      len(df_depto_atividades))

print("\nDistribuição por departamento:")

display(
    df_depto_atividades['Departamento_analise']
    .value_counts()
    .rename_axis('Departamento')
    .reset_index(name='Participantes')
)
```

```python

registros_departamentos = []

for departamento, grupo in df_depto_atividades.groupby(
    'Departamento_analise',
    sort=False
):
    for coluna in colunas_atividades:

        dados = grupo[coluna]
        base = int(dados.notna().sum())

        quantidade = int(dados.eq(True).sum())

        registros_departamentos.append({
            'Departamento': departamento,
            'Atividade': rotulos_atividades[coluna],
            'Participantes': quantidade,
            'Base': base,
            'Percentual': round(
                quantidade / base * 100, 1
            ) if base else 0
        })

df_departamento_atividades = pd.DataFrame(
    registros_departamentos
)

ordem_departamentos = (
    df_depto_atividades['Departamento_analise']
    .value_counts()
    .index
)

ordem_atividades = [
    rotulos_atividades[c]
    for c in colunas_atividades
]

tabela_depto_quantidade = (
    df_departamento_atividades
    .pivot(
        index='Departamento',
        columns='Atividade',
        values='Participantes'
    )
    .reindex(
        index=ordem_departamentos,
        columns=ordem_atividades
    )
    .fillna(0)
    .astype(int)
)

tabela_depto_percentual = (
    df_departamento_atividades
    .pivot(
        index='Departamento',
        columns='Atividade',
        values='Percentual'
    )
    .reindex(
        index=ordem_departamentos,
        columns=ordem_atividades
    )
)

print("QUANTIDADE DE PARTICIPANTES POR ATIVIDADE:")
display(tabela_depto_quantidade)

print("\nPERCENTUAL DENTRO DE CADA DEPARTAMENTO:")
display(tabela_depto_percentual)
```

```python

# Criar cópia de trabalho
df_etapa810 = df_cruzamentos[
    ['fe_familiaridade_ia', 'praticas_ensino_categoria']
].copy()

# Padronizar textos
for coluna in df_etapa810.columns:
    df_etapa810[coluna] = (
        df_etapa810[coluna]
        .astype('string')
        .str.strip()
    )

# Agrupar familiaridade
mapa_grupos = {
    'Nunca utilizei': 'Uso baixo ou nenhum',
    'Já utilizei, mas muito pouco': 'Uso baixo ou nenhum',
    'Utilizo com frequência para tarefas específicas':
        'Uso frequente ou diário',
    'Utilizo diariamente e integro em meu fluxo de trabalho':
        'Uso frequente ou diário'
}

df_etapa810['Grupo_familiaridade'] = (
    df_etapa810['fe_familiaridade_ia'].map(mapa_grupos)
)

# Verificar categorias não reconhecidas
if df_etapa810['Grupo_familiaridade'].isna().any():
    print("Categorias não reconhecidas:")
    display(
        df_etapa810.loc[
            df_etapa810['Grupo_familiaridade'].isna(),
            'fe_familiaridade_ia'
        ].value_counts(dropna=False)
    )
    raise ValueError("Verifique as categorias de familiaridade.")

# Conferir respostas ausentes
print("Respostas ausentes:")
print(df_etapa810.isna().sum())

df_etapa810_valido = df_etapa810.dropna(
    subset=['Grupo_familiaridade', 'praticas_ensino_categoria']
).copy()

print("\nParticipantes válidos:", len(df_etapa810_valido))

print("\nTamanho dos grupos:")
display(
    df_etapa810_valido['Grupo_familiaridade']
    .value_counts()
)
```

```python

grupos = [
    'Uso baixo ou nenhum',
    'Uso frequente ou diário'
]

# Tabela de quantidade
tabela_familiaridade_ensino = pd.crosstab(
    df_etapa810_valido['Grupo_familiaridade'],
    df_etapa810_valido['praticas_ensino_categoria']
).reindex(index=grupos, fill_value=0)

# Organizar categorias pela frequência geral
ordem_categorias = (
    tabela_familiaridade_ensino
    .sum(axis=0)
    .sort_values(ascending=False)
    .index
)

tabela_familiaridade_ensino = (
    tabela_familiaridade_ensino[ordem_categorias]
)

# Percentuais dentro de cada grupo
percentuais_familiaridade_ensino = (
    tabela_familiaridade_ensino
    .div(tabela_familiaridade_ensino.sum(axis=1), axis=0)
    .mul(100)
    .round(1)
)

print("QUANTIDADE DE PARTICIPANTES:")
display(tabela_familiaridade_ensino)

print("\nPERCENTUAL DENTRO DE CADA GRUPO:")
display(percentuais_familiaridade_ensino)

print("\nTOTAL POR GRUPO:")
display(tabela_familiaridade_ensino.sum(axis=1))
```

```python

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Patch

AZUL = '#2563EB'
CINZA = '#64748B'
CINZA_CLARO = '#E2E8F0'

grupos = [
    'Uso baixo ou nenhum',
    'Uso frequente ou diário'
]

qtd = tabela_familiaridade_ensino.reindex(
    index=grupos
)

perc = percentuais_familiaridade_ensino.reindex(
    index=grupos
)

# Rótulos mais curtos
rotulos = {
    'Uso crítico e transparente':
        'Uso crítico e transparente',

    'IA como conteúdo de estudo':
        'IA como conteúdo de estudo',

    'Preferência por múltiplas práticas':
        'Preferência por múltiplas práticas',

    'IA como ferramenta auxiliar docente':
        'IA como ferramenta auxiliar docente',

    'Restrição de IA em avaliações':
        'Restrição de IA em avaliações',

    'Uso crítico e transparente com ressalvas':
        'Uso crítico com ressalvas',

    'Adaptação de avaliações para limitar IA':
        'Adaptação das avaliações',

    'Crítica ao formato da pergunta':
        'Crítica ao formato da pergunta',

    'Necessidade de reflexão crítica':
        'Necessidade de reflexão crítica',

    'Posicionamento contrário à IA no ensino':
        'Posicionamento contrário à IA',

    'Sem opinião formada':
        'Sem opinião formada',

    'Uso orientado e limitado à pós-graduação':
        'Uso limitado à pós-graduação'
}

categorias = list(qtd.columns)

# Ordenar pela frequência geral
categorias = (
    qtd.sum(axis=0)
    .sort_values(ascending=True)
    .index.tolist()
)

qtd = qtd[categorias]
perc = perc[categorias]

y = np.arange(len(categorias))
altura = 0.34

fig, ax = plt.subplots(figsize=(14, 9))

barras_baixo = ax.barh(
    y - altura/2,
    perc.loc[grupos[0]].values,
    height=altura,
    color=CINZA
)

barras_frequente = ax.barh(
    y + altura/2,
    perc.loc[grupos[1]].values,
    height=altura,
    color=AZUL
)

# Identificação das barras
for barras, grupo in [
    (barras_baixo, grupos[0]),
    (barras_frequente, grupos[1])
]:
    base = int(qtd.loc[grupo].sum())

    for barra, categoria in zip(barras, categorias):
        quantidade = int(qtd.loc[grupo, categoria])
        percentual = perc.loc[grupo, categoria]

        texto = (
            f'{quantidade}/{base} | {percentual:.1f}%'
        ).replace('.', ',')

        ax.text(
            barra.get_width() + 0.8,
            barra.get_y() + barra.get_height()/2,
            texto,
            va='center',
            fontsize=9,
            color='#334155'
        )

ax.set_yticks(y)
ax.set_yticklabels(
    [rotulos.get(c, c) for c in categorias],
    fontsize=9
)

ax.set_xlim(0, 70)
ax.set_xticks([0, 10, 20, 30, 40, 50, 60, 70])
ax.set_xticklabels(
    ['0%', '10%', '20%', '30%', '40%', '50%', '60%', '70%']
)

ax.set_xlabel(
    'Percentual dentro de cada grupo de familiaridade',
    fontsize=10,
    labelpad=12
)

ax.grid(axis='x', linestyle='--', alpha=0.18)
ax.set_axisbelow(True)

for lado in ['top', 'right', 'left']:
    ax.spines[lado].set_visible(False)

ax.spines['bottom'].set_color(CINZA_CLARO)
ax.tick_params(axis='y', length=0, pad=10)

fig.suptitle(
    'Familiaridade com IA e Posicionamentos sobre o Ensino',
    fontsize=16,
    fontweight='bold',
    y=0.97
)

fig.legend(
    handles=[
        Patch(facecolor=CINZA, label='Uso baixo ou nenhum (n=30)'),
        Patch(facecolor=AZUL, label='Uso frequente ou diário (n=41)')
    ],
    loc='upper center',
    bbox_to_anchor=(0.5, 0.91),
    ncol=2,
    frameon=False,
    fontsize=9
)

fig.subplots_adjust(
    left=0.32,
    right=0.96,
    top=0.84,
    bottom=0.13
)

fig.text(
    0.32,
    0.055,
    'Base: 71 participantes. A pergunta permitia uma única resposta.',
    fontsize=9,
    color=CINZA
)

plt.show()
```
